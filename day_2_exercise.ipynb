## Strings

1. Given string contains a combination of the lower and upper case letters. Write a program to arrange the characters of a string so that all lowercase letters should come first.

```python
def arrange_letters(str) :
  lowercase = ""
  uppercase = ""
  for i in str :
    if i.islower() :
      lowercase += i
    else :
      uppercase += i
  return lowercase + uppercase

print(arrange_letters("saBrINa"))
```

2. Write a program to check if two strings are balanced. For example, strings s1 and s2 are balanced if all the characters in the s1 are present in s2. The character's position doesn't matter

```python
def is_balanced(s1, s2) :
  for i in s1 :
    if i not in s2 :
      return False
  return True
```

```python
def is_balanced(s1, s2) :
  for i in s1 :
    if i in s2 :
      return True
    else :
      return False

print(is_balanced("Kr", "SabrinaKreyz"))
```

```python
print(is_balanced("Kr", "Sabrina"))
```

2. Write a program to accept two integer numbers from the user and calculate their product (multiplication).

```python
def calculate_product() :
  first_number = int(input("Enter the first number."))
  second_number = int(input("Enter the second number."))
  print(f"The product of {first_number} and {second_number} is {first_number * second_number}")
```

```python
calculate_product()
```

```python
def display_product() :
  first_number = int(input("Enter the first number."))
  second_number = int(input("Enter the second number."))
  print(f"{first_number * second_number}")
```

```python
display_product()
```
