## Matplotlib

```python
import matplotlib.pyplot as plt
```

1. **Bar Plot.** Create a bar chart showing each person’s Name on the x-axis and their Score on the y-axis.

```python
df = pd.DataFrame

pd.DataFrame({
    "Names" : ['Anna', 'Ben', 'Cara', 'David'],
    "Ages" : [23, 35, 29, 41],
    "Scores" : [88, 92, 79, 85]
})
```

```python
bar_fig, bar_ax = plt.subplots()

bar_ax.bar(df["Names"], df["Scores"])
bar_ax.set_title("Test Scores by Student")
bar_ax.set_xlabel("Name")
bar_ax.set_ylabel("Score")

plt.show()
```

2. **Scatter Plot.** Create a scatter plot for df showing the relationship between Age and Score.

```python
fig, ax = plt.subplots()

ax.scatter(df["Ages"], df["Scores"],
marker = "s", # squares
facecolor = '#fb1', # orange
edgecolors = 'k', label = 'Weather Delays')

ax.set_title("Relationship between Age and Score")
ax.set_xlabel("Age")
ax.set_ylabel("Score")
plt.show()

plt.show()
```

3. **Histogram.** Create a histogram to show the distribution of the Score column.

```python
df["Scores"].hist(bins=5, edgecolor='black')

plt.title("Distribution of Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")

plt.show()
```
